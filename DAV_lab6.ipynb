```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# 1. Load dataset with encoding fix
df = pd.read_csv('/content/DAV FINAL PRODUCT BASED PROJECT DATASET.csv', encoding='latin1')

# 2. Parse Order_Date and extract year
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed', errors='coerce')
df['year'] = df['Order_Date'].dt.year
df['Postal_Code'] = df['Postal_Code'].fillna(0)
df_sorted = df.dropna(subset=['year']).sort_values('year')
df_sorted['year'] = df_sorted['year'].astype(int)

# --- Task 1: Core Interactive Charts ---
# 1. Univariate histogram of Sales with a range slider
fig1 = px.histogram(df, x="Sales", title="Task 1.1: Sales Distribution Histogram")
fig1.update_layout(xaxis_rangeslider_visible=True)
fig1.show()

# 2. Multivariate scatter plot (Sales vs Postal Code) with log_y scale and hover info
fig2 = px.scatter(df, x="Sales", y="Postal_Code", color="Category",
                  hover_data=["Customer_Name", "City", "State", "Order_Date"],
                  log_y=True,
                  title="Task 1.2: Sales vs Postal Code (Log Scale Y)")
fig2.show()

# 3. Box plot of Sales grouped by Category with all data points overlaid
fig3 = px.box(df, x="Category", y="Sales", points="all",
              title="Task 1.3: Sales Distribution by Product Category")
fig3.show()

# 4. 3D scatter plot (Sales, Postal Code, Year)
fig4 = px.scatter_3d(df_sorted, x="Sales", y="Postal_Code", z="year", color="Category",
                     title="Task 1.4: 3D Scatter (Sales, Postal Code, Year)")
fig4.show()

# --- Task 2: Animation and Dropdown Controls ---
# 1. Animated scatter plot across order years
fig5 = px.scatter(df_sorted, x="Sales", y="Postal_Code", animation_frame="year",
                  color="Category", size="Sales",
                  range_x=[0, df['Sales'].max() + 500],
                  range_y=[0, df['Postal_Code'].max() + 5000],
                  title="Task 2.1: Animated Sales Performance by Year")
fig5.show()

# 2. Chart with dropdown menu to switch the Y-axis metric dynamically
fig6 = go.Figure()
fig6.add_trace(go.Scatter(x=df['Sales'], y=df['Postal_Code'], mode='markers', name='Postal Code', visible=True))
fig6.add_trace(go.Scatter(x=df['Sales'], y=df['Row_ID'], mode='markers', name='Row ID', visible=False))

fig6.update_layout(
    title="Task 2.2: Sales vs Selectable Metric",
    xaxis_title="Sales",
    yaxis_title="Postal Code",
    updatemenus=[
        dict(
            active=0,
            buttons=list([
                dict(label="Postal Code", method="update", args=[{"visible": [True, False]}, {"yaxis": {"title": "Postal Code"}}]),
                dict(label="Row ID", method="update", args=[{"visible": [False, True]}, {"yaxis": {"title": "Row ID"}}]),
            ]),
            x=1.15, xanchor="right", y=1.1, yanchor="top"
        )
    ]
)
fig6.show()

# Written comparison comment
"""
Comparison Comment:
A static univariate view of sales provides simple summary counts but hides temporal trends.
The animated multivariate view reveals shift patterns in product category performance
across consecutive sales years that static views cannot capture.
"""
```
